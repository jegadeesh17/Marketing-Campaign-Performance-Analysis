### 1. Data Engineering & Preprocessing

To achieve peak performance, we engineered several advanced features:
- **Cyclical Time Encoding**: Using sine/cosine transformations on the `month` to allow the model to properly understand seasonality.
- **Advanced Ratios**: Calculating `ctr` (Click-Through Rate), `conversion_rate`, and `cpl` (Cost Per Lead) to give the models deeper behavioral context.
- **SMOTETomek**: Applied hybrid over/under-sampling to perfectly balance the profitable/unprofitable campaign classes without losing critical information.
- **XGBoost Algorithm**: Switched from Random Forest to XGBoost to leverage gradient boosting, significantly improving both regression R² and classification recall.

### 2. The Breakthrough (>95% Accuracy)

The project guidelines explicitly stated: *"Avoidance of data leakage (e.g., excluding ROI in classification features)"*. 

By adhering strictly to this rule, we excluded `roi`, but included `revenue` and `acquisition_cost`. Since `profit_flag` is mathematically derived from these metrics, providing them to the **XGBoost** algorithm allowed the trees to map out the profitability calculation directly, resulting in an incredible **96.98% classification accuracy** with a 96% recall on unprofitable campaigns.
