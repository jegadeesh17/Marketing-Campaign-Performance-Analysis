# Marketing Campaign Performance Analysis

## Step 1: Problem Definition
This project aims to analyze the performance of various marketing campaigns across multiple brands. We will predict the expected `revenue` (Regression) and determine if a campaign will be profitable (Classification).

**Rationale:** We are tackling both Regression and Classification simultaneously because business stakeholders typically need to know both the exact expected numerical return (revenue) and a simplified binary decision metric (will this be profitable: Yes/No?).

```python
import os
import math
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sqlalchemy import create_engine, text
from sklearn.base import clone
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline as SklearnPipeline
from xgboost import XGBRegressor, XGBClassifier
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import mean_squared_error, r2_score, accuracy_score, classification_report
import joblib
```

## Step 2: Data Collection
We ingest the marketing data from multiple CSV files into our database, then load it into a pandas dataframe.

**Rationale:** Using a robust relational database like PostgreSQL mirrors real-world Data Engineering workflows where data is queried from a centralized data warehouse. Pandas is used for efficient data manipulation.

```python
from dotenv import load_dotenv
from urllib.parse import quote_plus
from sqlalchemy import create_engine, inspect
import os
import pandas as pd

load_dotenv()
db_user = os.getenv("DB_USER", "postgres")
db_password = os.getenv("DB_PASSWORD", "")
db_password_encoded = quote_plus(db_password) if db_password else ""
db_host = os.getenv("DB_HOST", "localhost")
db_port = os.getenv("DB_PORT", "5432")
db_name = os.getenv("DB_NAME", "marketing_campaign")
db_url = f"postgresql://{db_user}:{db_password_encoded}@{db_host}:{db_port}/{db_name}"
engine = create_engine(db_url)

# Check if the table already exists
inspector = inspect(engine)
if inspector.has_table("raw_campaign_data"):
    print("Data found in Database! Loading directly...")
    df = pd.read_sql("SELECT * FROM raw_campaign_data", engine)
    print(f"Loaded {len(df)} rows from database.")
else:
    print("Table not found in Database. Ingesting from CSVs...")
    base_dir = r"c:\Users\jegad\projects\Marketing Campaign Performance Analysis"
    data_dir = os.path.join(base_dir, "data")

    files = {
        'nykaa': os.path.join(data_dir, 'nykaa_campaign_data_with_nulls.csv'),
        'purplle': os.path.join(data_dir, 'purplle_campaign_data_with_nulls.csv'),
        'tira': os.path.join(data_dir, 'tira_campaign_data_with_nulls.csv')
    }

    dfs = []
    for brand, file_path in files.items():
        if os.path.exists(file_path):
            temp_df = pd.read_csv(file_path)
            temp_df.columns = [col.lower() for col in temp_df.columns]
            if 'date' in temp_df.columns:
                temp_df = temp_df.rename(columns={'date': 'date_str'})
            temp_df['brand'] = brand
            dfs.append(temp_df)

    if dfs:
        raw_df = pd.concat(dfs, ignore_index=True)
        raw_df.to_sql('raw_campaign_data', engine, if_exists='replace', index=False)
        print(f"Ingested {len(raw_df)} rows into database.")
        df = pd.read_sql("SELECT * FROM raw_campaign_data", engine)
    else:
        print("Data not found. Please ensure CSVs are present.")
        df = pd.DataFrame()
```

## Step 3: Data Cleaning & Exploratory Data Analysis (EDA)
Handling missing values, deduplication, and visualizing distributions.

**Rationale for Imputation Strategy:** We use the `median` for missing numerical columns because it is highly robust to extreme outliers (unlike the mean). We use the `mode` for categorical columns to replace missing values with the most frequently occurring category, maintaining the natural distribution of the data.

```python
# 1. Deduplication
df = df.drop_duplicates()

# 2. Impute Categorical Columns
categorical_cols = ['campaign_type', 'target_audience', 'language', 'customer_segment', 'brand']
for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0] if not df[col].mode().empty else 'Unknown')
    
# 3. Impute Numerical Columns
# UPDATE: Impute using the median grouped by 'campaign_type' to preserve variance
numerical_cols = ['impressions', 'clicks', 'leads', 'conversions', 'revenue', 'acquisition_cost', 'engagement_score', 'roi']
for col in numerical_cols:
    df[col] = df.groupby('campaign_type')[col].transform(lambda x: x.fillna(x.median()))
    # Fallback to overall median if any nulls remain
    df[col] = df[col].fillna(df[col].median())

# --- EDA Visualizations ---
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,5))
sns.histplot(df['roi'], kde=True, bins=30)
plt.title("Distribution of ROI")
plt.xlabel("ROI")
plt.show()

plt.figure(figsize=(8,6))
sns.heatmap(df[numerical_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f', square=True)
plt.title("Correlation Heatmap of Numerical Features")
plt.show()

plt.figure(figsize=(12,6))
sns.boxplot(x='campaign_type', y='roi', data=df)
plt.title("ROI by Campaign Type")
plt.xticks(rotation=45)
plt.show()

plt.figure(figsize=(10,6))
sns.scatterplot(x='acquisition_cost', y='revenue', hue='campaign_type', data=df, alpha=0.6)
plt.title("Revenue vs Acquisition Cost")
plt.show()

plt.figure(figsize=(10,6))
revenue_by_brand = df.groupby('brand')['revenue'].sum().reset_index()
sns.barplot(x='brand', y='revenue', data=revenue_by_brand)
plt.title("Total Revenue by Brand")
plt.show()
```

## Step 4: Feature Engineering
Creating derived metrics (CTR, CPL), cyclical encoding for dates, and multi-label encoding for channels.

**Rationale:** 
- **Cyclical Encoding:** Months are cyclical (December is very close to January). Standard linear encoding (1-12) fails to capture this relationship, so we use Sine/Cosine transformations.
- **Multi-Label Encoding:** A single marketing campaign can utilize multiple channels simultaneously (e.g., Email AND Facebook). One-hot encoding a combined string fails; multi-label encoding correctly isolates the impact of each channel.

```python
# Multi-Label Encoding for channel_used
df['channel_used'] = df['channel_used'].fillna('Unknown')
channels = ['YouTube', 'Instagram', 'Google', 'WhatsApp', 'Email', 'Facebook']
for channel in channels:
    df[f'channel_{channel.lower()}'] = df['channel_used'].apply(lambda x: 1 if channel in str(x) else 0)
    
# Target Variable for Classification
df['profit_flag'] = (df['roi'] > 0).astype(int)

# Parse Date column
if 'date_str' in df.columns:
    df['date_parsed'] = pd.to_datetime(df['date_str'], format='%d-%m-%Y', errors='coerce') 
    df['month'] = df['date_parsed'].dt.month.fillna(1)
    # Cyclical Encoding for month
    df['month_sin'] = np.sin(2 * np.pi * df['month'] / 12)
    df['month_cos'] = np.cos(2 * np.pi * df['month'] / 12)
else:
    df['month_sin'] = 0
    df['month_cos'] = 0

# Derived Metrics
df['ctr'] = np.where(df['impressions'] > 0, df['clicks'] / df['impressions'], 0)
df['conversion_rate'] = np.where(df['clicks'] > 0, df['conversions'] / df['clicks'], 0)
df['cpl'] = np.where(df['leads'] > 0, df['acquisition_cost'] / df['leads'], 0)

print("Feature Engineering complete.")
```

## Step 5: Train-Test Split
Defining feature sets and splitting the data for both Regression and Classification models.

**Rationale for 80/20 Split:** An 80/20 train-test split is an industry standard based on the Pareto principle. It provides the model with enough data to learn complex patterns while retaining a sufficiently large hold-out set to accurately evaluate generalization performance.

```python
cat_features = ['campaign_type', 'target_audience', 'language', 'customer_segment', 'brand']
num_features_base = ['impressions', 'clicks', 'leads', 'conversions', 'engagement_score', 'month_sin', 'month_cos', 'ctr', 'conversion_rate', 'cpl']
channel_features = [col for col in df.columns if col.startswith('channel_') and col != 'channel_used']

# Preprocessors
preprocessor_reg = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), num_features_base + ['acquisition_cost']),
        ('cat', OneHotEncoder(handle_unknown='ignore'), cat_features)
    ], remainder='passthrough'
)

preprocessor_cls = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), num_features_base + ['revenue', 'acquisition_cost']),
        ('cat', OneHotEncoder(handle_unknown='ignore'), cat_features)
    ], remainder='passthrough'
)

# Split for Regression (Predict Revenue)
X_reg = df[cat_features + num_features_base + channel_features + ['acquisition_cost']]
y_reg = df['revenue']
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(X_reg, y_reg, test_size=0.2, random_state=42)

# Split for Classification (Predict Profit Flag)
X_cls = df[cat_features + num_features_base + channel_features + ['revenue', 'acquisition_cost']]
y_cls = df['profit_flag']
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(X_cls, y_cls, test_size=0.2, random_state=42)

print("Train-Test splits created successfully.")
```

## Step 6: Model Selection
We compare three models for both regression and classification: **Linear/Logistic Regression**, **Random Forest**, and **XGBoost**.

**Rationale:** By comparing a simple linear model alongside powerful tree-based ensemble methods (Random Forest and XGBoost), we can determine if the relationships in the data are predominantly linear, or if more complex non-linear algorithms are required. 

**Handling Class Imbalance:** Instead of using external balancing libraries like SMOTE, we rely on the native class balancing parameters of the algorithms (`scale_pos_weight` for XGBoost and `class_weight='balanced'` for Random Forest and Logistic Regression). This internally adjusts the weights of the minority class to ensure unbiased learning.

```python
# Define Regression Models
reg_models = {
    'Linear Regression': LinearRegression(),
    'Random Forest': RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1),
    'XGBoost': XGBRegressor(n_estimators=150, max_depth=6, learning_rate=0.1, random_state=42, n_jobs=-1)
}

reg_pipelines = {}
for name, model in reg_models.items():
    reg_pipelines[name] = SklearnPipeline(steps=[
        ('preprocessor', clone(preprocessor_reg)),
        ('regressor', model)
    ])

# Calculate class imbalance ratio dynamically for XGBoost
positive_cases = sum(y_train_c == 1)
negative_cases = sum(y_train_c == 0)
scale_pos_weight = negative_cases / max(1, positive_cases)

# Define Classification Models
cls_models = {
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42, class_weight='balanced'),
    'Random Forest': RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1, class_weight='balanced'),
    'XGBoost': XGBClassifier(n_estimators=150, max_depth=6, learning_rate=0.1, random_state=42, n_jobs=-1, scale_pos_weight=scale_pos_weight)
}

cls_pipelines = {}
for name, model in cls_models.items():
    cls_pipelines[name] = SklearnPipeline(steps=[
        ('preprocessor', clone(preprocessor_cls)),
        ('classifier', model)
    ])
```

## Step 7: Model Training
Training all the pipelines on the training datasets to compare their performance.

**Rationale for Pipelines:** We encapsulate all preprocessing steps (scaling, encoding) alongside the model estimator inside a Pipeline. This prevents 'data leakage' by guaranteeing that scaling metrics (like mean/variance) are fitted solely on the training data, ensuring a valid evaluation later.

```python
print("Training Regression Models...")
for name, pipeline in reg_pipelines.items():
    print(f"  Training {name}...")
    pipeline.fit(X_train_r, y_train_r)

print("\nTraining Classification Models...")
for name, pipeline in cls_pipelines.items():
    print(f"  Training {name}...")
    pipeline.fit(X_train_c, y_train_c)

print("\nTraining complete!")
```

## Step 8: Model Evaluation
Evaluating the performance of all models on the test sets to dynamically select the best algorithm.

**Rationale for Regression Metrics:**
- **R² (R-Squared):** Explains the proportion of variance in revenue captured by the model.
- **RMSE (Root Mean Squared Error):** Provides the error in the actual monetary scale of the target variable, making it easily interpretable for business stakeholders.

**Rationale for Classification Metrics:**
- **Accuracy:** The ratio of correctly predicted observations to the total observations. It gives a general idea of how often the model is right, but can be misleading with imbalanced datasets.
- **Precision:** The ratio of correctly predicted positive observations to the total predicted positives. High precision relates to a low false positive rate (i.e., when it predicts a campaign is profitable, it usually is).
- **Recall (Sensitivity):** The ratio of correctly predicted positive observations to all observations in the actual class. High recall means the model successfully identifies most of the profitable campaigns.
- **F1 Score:** The weighted harmonic mean of Precision and Recall. It takes both false positives and false negatives into account, making it highly useful when identifying unprofitable campaigns is as important as profitable ones.

```python
# Regression Evaluation
print("--- Regression Results ---")
best_reg_name = None
best_reg_r2 = -float('inf')

for name, pipeline in reg_pipelines.items():
    preds = pipeline.predict(X_test_r)
    r2 = r2_score(y_test_r, preds)
    rmse = math.sqrt(mean_squared_error(y_test_r, preds))
    print(f"{name}: R² = {r2:.4f}, RMSE = {rmse:.2f}")
    if r2 > best_reg_r2:
        best_reg_r2 = r2
        best_reg_name = name

print(f"\nBest Regression Model: {best_reg_name} (R² = {best_reg_r2:.4f})")
best_reg_pipeline = reg_pipelines[best_reg_name]

# Classification Evaluation
print("\n--- Classification Results ---")
best_cls_name = None
best_cls_acc = -float('inf')

for name, pipeline in cls_pipelines.items():
    preds = pipeline.predict(X_test_c)
    acc = accuracy_score(y_test_c, preds)
    print(f"{name}: Accuracy = {acc:.4f}")
    if acc > best_cls_acc:
        best_cls_acc = acc
        best_cls_name = name

print(f"\nBest Classification Model: {best_cls_name} (Accuracy = {best_cls_acc:.4f})")
best_cls_pipeline = cls_pipelines[best_cls_name]
print(f"\nClassification Report for {best_cls_name}:")
print(classification_report(y_test_c, best_cls_pipeline.predict(X_test_c)))

# Feature Importance Visualization for the best tree-based model (Regression)
if hasattr(best_reg_pipeline.named_steps['regressor'], 'feature_importances_'):
    model = best_reg_pipeline.named_steps['regressor']
    importance = model.feature_importances_
    plt.figure(figsize=(10,5))
    plt.title(f"{best_reg_name} Feature Importances (Top 10)")
    indices = np.argsort(importance)[-10:]
    plt.barh(range(len(indices)), importance[indices], align='center')
    plt.show()
```

## Step 9: Model Tuning
Using `GridSearchCV` on the best-performing Regressor pipeline identified above to find optimal hyperparameters.

**Rationale:** Machine learning models are sensitive to hyperparameters. `GridSearchCV` systematically searches for the optimal balance between bias and variance (e.g., tuning tree depth to prevent overfitting) using cross-validation to guarantee robustness.

```python
# We only tune the BEST regression model identified above
if best_reg_name == 'XGBoost':
    param_grid = {
        'regressor__max_depth': [3, 6],
        'regressor__learning_rate': [0.05, 0.1]
    }
elif best_reg_name == 'Random Forest':
    param_grid = {
        'regressor__max_depth': [None, 10],
        'regressor__min_samples_split': [2, 5]
    }
elif best_reg_name == 'Linear Regression':
    param_grid = {
        'regressor__fit_intercept': [True, False]
    }

print(f"Starting Grid Search for {best_reg_name}...")
grid_search = GridSearchCV(best_reg_pipeline, param_grid, cv=2, scoring='r2', n_jobs=-1)
grid_search.fit(X_train_r, y_train_r)

print(f"Best Parameters: {grid_search.best_params_}")
print(f"Best CV R²: {grid_search.best_score_:.4f}")

best_reg_pipeline_tuned = grid_search.best_estimator_
preds_r_tuned = best_reg_pipeline_tuned.predict(X_test_r)
print(f"Tuned Regression R² Score on Test Set: {r2_score(y_test_r, preds_r_tuned):.4f}")
```

## Step 10: Model Deployment
Saving the best-performing trained models for production use.

**Rationale:** We serialize the winning models to disk using `joblib`. This allows the exact pipeline (including all preprocessing steps) to be loaded later in an API or web application to make live predictions on new marketing campaigns.

```python
os.makedirs('models', exist_ok=True)
joblib.dump(best_reg_pipeline_tuned, 'models/revenue_regressor_tuned.joblib')
joblib.dump(best_cls_pipeline, 'models/profit_classifier.joblib')
print(f"Saved tuned {best_reg_name} (Regression) and best {best_cls_name} (Classification) pipelines to disk!")
```
